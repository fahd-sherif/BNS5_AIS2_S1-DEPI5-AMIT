```python
class person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self): #user
        print('__str__called')
        return f"person (name= '{self.name}', age= {self.age})"
    def __repr__(self): #developer
        return self.name
```

```python
p1 = person("Fahd", 21)
```

```python
print(p1)
```

```python
str(p1)
```
