<h3>Two Sum</h3>

```python
class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        for i in range(0,len(nums)-1):
            for j in range(i+1,len(nums)):
                if nums[i] + nums[j] == target:
                    return [i,j]
```

```python
t1 = Solution()
print(t1.twoSum([2,7,11,15], 9))
```
