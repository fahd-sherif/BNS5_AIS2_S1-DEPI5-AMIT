```python
class MultiplicationTable:
    """
    MultiplicationTable class handles printing
    the multiplication table of a given number from 1 to 10.
    Attributes:
        number (int): The number to print its multiplication table.
    """
    def __init__(self, number):
        """
        Initialize the MultiplicationTable object.
        Args:
            number (int): The number for the multiplication table.
        """
        self.number = number
    def multiplication_table(self):
        """
        Prints the multiplication table of the stored number from 1 to 10.
        Returns:
            None
        """
        for i in range(1, 11):
            print(f"{self.number} * {i} = {self.number * i}")

num = int(input("Enter a number: "))


table = MultiplicationTable(num)
table.multiplication_table()
```

```python
class TwinPrimes:
    """
    TwinPrimes class handles checking prime numbers
    and printing all twin prime pairs below a given limit.
    Attributes:
        limit (int): The upper limit for searching twin primes.
    """
    def __init__(self, limit):
        """
        Initialize the TwinPrimes object.
        Args:
            limit (int): The upper limit for searching twin primes.
        """
        self.limit = limit
    def is_prime(self, number):
        """
        Checks whether a number is prime.
        Args:
            number (int): The number to check.
        Returns:
            bool: True if the number is prime, otherwise False.
        """
        if number < 2:
            return False
        for i in range(2, int(number ** 0.5) + 1):
            if number % i == 0:
                return False
        return True
    def print_twin_primes(self):
        """
        Prints all twin prime pairs less than the given limit.
        Twin primes are two prime numbers that differ by 2.
        Returns:
            None
        """
        print("Twin Primes:")
        
        for num in range(3, self.limit - 1, 2):
            if self.is_prime(num) and self.is_prime(num + 2):
                print(f"{num} and {num + 2}")


twin_primes = TwinPrimes(1000)
twin_primes.print_twin_primes()
```

```python
class PrimeFactors:
    """
    PrimeFactors class handles finding all prime factors
    of a given number.
    Attributes:
        number (int): The number to find its prime factors.
    """
    def __init__(self, number):
        """
        Initialize the PrimeFactors object.
        Args:
            number (int): The number to find its prime factors.
        """
        self.number = number
    def prime_factors(self):
        """
        Find all prime factors of the stored number.
        Returns:
            list: A list containing all prime factors of the number.
        """
        factors = []
        while self.number % 2 == 0:
            factors.append(2)
            self.number = self.number // 2
        i = 3
        while i * i <= self.number:
            while self.number % i == 0:
                factors.append(i)
                self.number = self.number // i
            i += 2
        if self.number > 2:
            factors.append(self.number)
        return factors


number = PrimeFactors(56)
result = number.prime_factors()
print(result)
```

```python
class DecimalToBinary:
    """
    DecimalToBinary class handles converting a decimal number
    to its binary representation.
    Attributes:
        number (int): The decimal number to convert.
    """
    def __init__(self, number):
        """
        Initialize the DecimalToBinary object.
        Args:
            number (int): The decimal number to convert.
        """
        self.number = number
    def dec_to_bin(self):
        """
        Convert the stored decimal number to its binary representation.
        Returns:
            str: The binary representation of the decimal number.
        """
        if self.number == 0:
            return "0"
        binary = ""
        while self.number > 0:
            binary = str(self.number % 2) + binary
            self.number = self.number // 2
        return binary



number = DecimalToBinary(11)
result = number.dec_to_bin()
print(result)
```

```python
class PerfectNumbers:
    """
    PerfectNumbers class handles printing all perfect numbers
    in a given range.
    Attributes:
        start (int): The starting number of the range.
        end (int): The ending number of the range.
    """
    def __init__(self, start, end):
        """
        Initialize the PerfectNumbers object.
        Args:
            start (int): The starting number of the range.
            end (int): The ending number of the range.
        """
        self.start = start
        self.end = end
    def perfect_nums(self):
        """
        Print all perfect numbers in the given range.
        Returns:
            None
        """
        for number in range(self.start, self.end + 1):
            if number == 0:
                print(0)
                continue
            total = 0
            for i in range(1, number):
                if number % i == 0:
                    total += i
            if total == number:
                print(number)



numbers = PerfectNumbers(0, 100)
numbers.perfect_nums()
```

convert calculator function to class

```python
class Calculator:
    """
    Calculator class handles basic arithmetic operations
    and provides an interactive calculator menu.
    Attributes:
        num1 (float): The first number.
        num2 (float): The second number.
    """
    def __init__(self):
        """
        Initialize the Calculator object.
        The numbers are initialized as None and will be
        assigned when the user enters them.
        """
        self.num1 = None
        self.num2 = None
    def add(self):
        """
        Add the two stored numbers.
        Returns:
            float: The sum of the two numbers.
        """
        return self.num1 + self.num2
    def subtract(self):
        """
        Subtract the second number from the first.
        Returns:
            float: The difference between the two numbers.
        """
        return self.num1 - self.num2
    def multiply(self):
        """
        Multiply the two stored numbers.
        Returns:
            float: The product of the two numbers.
        """
        return self.num1 * self.num2
    def divide(self):
        """
        Divide the first number by the second.
        Raises:
            ZeroDivisionError: If the second number is zero.
        Returns:
            float: The division result.
        """
        if self.num2 == 0:
            raise ZeroDivisionError("Cannot divide by zero.")
        return self.num1 / self.num2
    def get_number(self, message):
        """
        Get a valid number from the user.
        Args:
            message (str): The message displayed to the user.
        Returns:
            float: A valid floating-point number.
        """
        while True:
            try:
                return float(input(message))
            except ValueError:
                print("Invalid input! Please enter a valid number.")
    def calculator(self):
        """
        Run the calculator program.
        Displays a menu, performs the selected operation,
        and allows the user to repeat calculations until
        they choose to exit.
        Returns:
            None
        """
        while True:
            print("\n===== Basic Calculator =====")
            print("1. Addition")
            print("2. Subtraction")
            print("3. Multiplication")
            print("4. Division")
            print("Type 'exit' to quit.")
            choice = input(
                "Choose an operation (1-4) or type 'exit': "
            ).lower()
            if choice == "exit":
                print("Thank you for using the calculator!")
                break
            if choice not in ["1", "2", "3", "4"]:
                print("Invalid choice! Please try again.")
                continue
            self.num1 = self.get_number("Enter the first number: ")
            self.num2 = self.get_number("Enter the second number: ")
            try:
                if choice == "1":
                    result = self.add()
                    print(
                        f"The result of adding "
                        f"{self.num1} and {self.num2} is {result}"
                    )
                elif choice == "2":
                    result = self.subtract()
                    print(
                        f"The result of subtracting "
                        f"{self.num2} from {self.num1} is {result}"
                    )
                elif choice == "3":
                    result = self.multiply()
                    print(
                        f"The result of multiplying "
                        f"{self.num1} and {self.num2} is {result}"
                    )
                elif choice == "4":
                    result = self.divide()
                    print(
                        f"The result of dividing "
                        f"{self.num1} by {self.num2} is {result}"
                    )
            except ZeroDivisionError as error:
                print(error)
            again = input(
                "\nDo you want another calculation? (yes/no): "
            ).lower()
            if again != "yes":
                print("Thank you for using the calculator!")
                break


calculator = Calculator()
calculator.calculator()
```

convert thanos project from functions to class

```python
import os
import random


class ThanosProject:
    """
    ThanosProject class handles creating a folder,
    generating text files, and randomly deleting half
    of the files.
    Attributes:
        folder_path (str): The name or path of the folder to create.
        num_files (int): The number of text files to generate.
    """
    def __init__(self, folder_path: str, num_files: int) -> None:
        """
        Initialize the ThanosProject object.
        Args:
            folder_path (str): The name or path of the folder to create.
            num_files (int): The number of text files to generate.
        """
        self.folder_path = folder_path
        self.num_files = num_files
    def create_files(self) -> None:
        """
        Create the folder and generate text files.
        Returns:
            None
        """
        if not os.path.exists(self.folder_path):
            os.mkdir(self.folder_path)
        for i in range(1, self.num_files + 1):
            file_name = f"file_{i}.txt"
            file_path = os.path.join(self.folder_path, file_name)
            with open(file_path, "w") as file:
                file.write(f"This is file number {i}")
    def delete_half_files(self) -> None:
        """
        Randomly delete half of the files in the folder.
        Returns:
            None
        """
        files = os.listdir(self.folder_path)
        print("Files before delete:")
        print(files)
        print("Number of files:", len(files))
        random_files = random.sample(files, len(files) // 2)
        for file in random_files:
            os.remove(os.path.join(self.folder_path, file))
        remaining_files = os.listdir(self.folder_path)
        print("\nDeleted files:")
        print(random_files)
        print("\nRemaining files:")
        print(remaining_files)
        print("Number of remaining files:", len(remaining_files))
    def thanos_project(self) -> None:
        """
        Run the complete Thanos Project.
        Creates the files and then randomly deletes
        half of them.
        Returns:
            None
        """
        self.create_files()
        self.delete_half_files()


folder: str = input("Enter folder name: ")
number: int = int(input("Enter number of files: "))

project = ThanosProject(folder, number)
project.thanos_project()
```
