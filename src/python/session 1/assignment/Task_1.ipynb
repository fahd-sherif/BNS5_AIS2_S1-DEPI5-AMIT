<h4>Task 1</h4>

```python
Email = "Amit_ml@gmail.edu"
#validation
if Email.count("@")!= 1 or "." not in Email.split("@")[1]:
    print("InValid Email")
else:
    print("Valid Email")
```

```python
#Extract Username
Username = Email.split("@")[0]
Username
```

```python
# Extract Domain
Email_c = Email.split("@")
Domain = Email_c[1].split(".")
Domain[0]
```

```python
#Check for Domain Ending
if Email.endswith(".com"):
    print("Commercial Domain")
elif Email.endswith(".edu"):
    print("Educational Domain")
else:
    print("Other Domain")
```
