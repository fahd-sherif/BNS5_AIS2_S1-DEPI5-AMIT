<p>1. Write a function that inputs a number and prints the multiplication table of that number </p>

```python
def multiplication_table(number):
    """
    Prints the multiplication table of a given number from 1 to 10.
    param number: The number to print its multiplication table.
    type number: int
    return: None
    rtype: None
    """
    for i in range(1, 11):
        print(f"{number} * {i} = {number * i}")
num = int(input("Enter a number: "))
multiplication_table(num)
```

<p>2. Write a program to print twin primes less than 1000. If two consecutive odd numbers are both prime then they are known as twin primes</p>

```python
def is_prime(number):
    """
    Checks whether a number is prime.
    param number: The number to check.
    type number: int
    return: True if the number is prime, otherwise False.
    rtype: bool
    """
    if number < 2:
        return False
    for i in range(2, int(number ** 0.5) + 1):
        if number % i == 0:
            return False
    return True
def print_twin_primes(limit):
    """
    Prints all twin prime pairs less than the given limit.
    Twin primes are two prime numbers that differ by 2.
    param limit: The upper limit for searching twin primes.
    type limit: int
    return: None
    rtype: None
    """
    print("Twin Primes:")
    for num in range(3, limit - 1, 2):
        if is_prime(num) and is_prime(num + 2):
            print(f"{num} and {num + 2}")
print_twin_primes(1000)
```

<p>3. Write a program to find out the prime factors of a number. Example: prime factors of 56 - 2, 2, 2, 7 </p>

```python
def primeFactors(number):
    """
    Find all prime factors of a given number.
    param number: The number to find its prime factors.
    type number: int
    return: A list containing all prime factors of the number.
    rtype: list
    """
    factors = []
    while number % 2 == 0:
        factors.append(2)
        number = number // 2
    i = 3
    while i * i <= number:
        while number % i == 0:
            factors.append(i)
            number = number // i
        i += 2
    if number > 2:
        factors.append(number)
    return factors
print(primeFactors(56))
```

<p>4. Write a function that converts a decimal number to binary number </p>

```python
def decToBin(number):
    """
    Convert a decimal number to its binary representation.
    param number: The decimal number to convert.
    type number: int
    return: The binary representation of the decimal number.
    rtype: str
    """
    if number == 0:
        return "0"
    binary = ""
    while number > 0:
        binary = str(number % 2) + binary
        number = number // 2
    return binary
print(decToBin(11))
```

<p>5. A number is called perfect if the sum of proper divisors of that number is equal to the number. For example 28 is perfect number, since 1+2+4+7+14=28. Write a program to print all the perfect numbers in a given range </p>

```python
def perfectNums(start, end):
    """
    Print all perfect numbers in a given range.
    param start: The starting number of the range.
    type start: int
    param end: The ending number of the range.
    type end: int
    return: None
    rtype: None
    """
    for number in range(start, end + 1):
        if number == 0:
            print(0)
            continue
        total = 0
        for i in range(1, number):
            if number % i == 0:
                total += i
        if total == number:
            print(number)
perfectNums(0, 100)
```
