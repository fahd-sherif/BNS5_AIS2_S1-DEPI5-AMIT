```python
def add(x:float, y:float):
    '''div function
    args:
        param_1 = user must input the first num
        type param_1 = float
        param_2 = user must input the second num
        type param_1 = float
        return x+y
    '''
    return x+y
def  sub(x:float, y:float):
    '''div function
        args:
            param_1 = user must input the first num
            type param_1 = float
            param_2 = user must input the second num
            type param_1 = float
            return x-y
        '''
    return x-y
def  mult(x:float, y:float):
    '''div function
        args:
            param_1 = user must input the first num
            type param_1 = float
            param_2 = user must input the second num
            type param_1 = float
            return x*y
        '''
    return x*y
def div(x:float, y:float):
    '''div function
        args:
            param_1 = user must input the first num
            type param_1 = float
            param_2 = user must input the second num
            type param_1 = float
            return x/y
        '''
    return x/y
```

```python
def factorial(num:int):
    '''  
    Calculate n! using recursion.
    
    Parameters
    ----------
    num : int
    
    Returns
    -------
    int
    '''
    if num <0:
        raise ValueError("Factorial is undefined negative values")
    if num == 0 or num == 1:
        return 1

    return num * factorial(num -1)
```

```python
factorial(3)
```

```python
def is_prime(num:int)->bool:
    '''  
    Check whether a number is prime.
    
    Parameters
    ----------
    num : int
    Number to test.
    
    Returns
    -------
    bool
    True if prime, otherwise False.
    
    '''

    if num < 2:
        return False
    for i in range(2,num):
        if num % i == 0:
            return False
    return True

```

```python
is_prime(23)
```

```python
def generate_password(length: int) -> str:
    char = "absshsjskskdsfkdfksdfkjdhfhdjfhkjdkfuhkdfuhekfjdfks$%^@#!@#$"
    password = ""

    for _ in range(length):
        password += random.choice(char)
    return password
```

```python
import string
import secrets
def generate_password_secure(length:int =12):
    if length <=0:
        raise ValueError ("Password length must be more than zero")
    char = (
        string.ascii_letters +
        string.digits +
        string.punctuation
    )
    return "".join(secrets.choice (char) for _ in range(length))
generate_password_secure(12)
```

```python
words = [
    "welcome", "Ali", "hi", "Ali", "Hello", "Ali", "hi", "Ali",
    "Hello", "Ali", "hi", "Ali", "Hello", "Ali", "hi", "Ali",
    "Hello", "Ali", "hi", "Ali", "Hello", "Ali", "hi", "Ali",
    "Hello", "Ali", "hi", "Ali"
]
def words_count(words: list) -> dict:
    """
    Count the occurrences of each unique word in a list.

    Parameters
    ----------
    words : list
        List of words to count.

    Returns
    -------
    dict
        Dictionary with unique words as keys and their counts as values.
    """

    unique_words = set(words)
    count_dict = {word: words.count(word) for word in unique_words}
    for i in unique_words:
        for j in words:
            if i == j:
                count_dict[i] += 1

    return count_dict
print(words_count(words))
```

```python
import random

# Predefined responses
responses = {
    "hello": ["Hello!", "Hi there!", "Greetings!"],
    "how are you": ["I'm doing well, thank you!", "I'm fine, how about you?"],
    "goodbye": ["Goodbye!", "See you later!", "Farewell!"],
    "default": ["I'm sorry, I didn't understand.", "Could you please rephrase that?"]
}

# Function to get a response based on user input
def get_response(user_input):
    for key in responses:
        if key in user_input:
            return random.choice(responses[key])
    return random.choice(responses["default"])

# Chatbot function
def chatbot():
    print("Chatbot: Hi! How can I assist you today?")

    while True:
        user_input = input("User: ").lower()
        response = get_response(user_input)
        print("Chatbot:", response)

        if user_input == "goodbye":
            break

# Run the chatbot
chatbot()
```
